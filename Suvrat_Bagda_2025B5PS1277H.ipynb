```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation # Specifically for creating animations, like the 'live' depth monitor.
import numpy as np
import pandas as pd
```

```python
from pathlib import Path

#  the URL for the raw CSV data on GitHub.
data_path = "https://raw.githubusercontent.com/suvratprofessional-hub/odyssesus_journey/refs/heads/main/Depth%20Data.csv"

#  CSV data into a pandas DataFrame.
df = pd.read_csv(data_path)

# Rename columns for better readability and consistency
df = df.rename(columns={'Point': 'point', 'Depth (m)': 'depth'})


# Convert 'point' and 'depth' columns to numeric types. 'coerce' will turn invalid parsing into NaN.
df['point'] = pd.to_numeric(df['point'], errors='coerce')
df['depth'] = pd.to_numeric(df['depth'], errors='coerce')

# Drop rows where 'point' or 'depth' values are NaN (Not a Number)
# .copy() is used to avoid SettingWithCopyWarning in future operations.
df = df.dropna(subset=['point', 'depth']).copy()

# --- Outlier Identification and Removal, and Data Smoothing ---

col = 'depth'#using quartile mtd for outlier detetction
# Calculate the first quartile (Q1) and third quartile (Q3) for outlier detection.
q25 = df[col].quantile(0.25)
q75 = df[col].quantile(0.75)

# Calculate the Interquartile Range (IQR).
iqr = q75 - q25
# Determine the lower and upper bounds for outlier detection using the 1.5*IQR rule.
min_bound = q25 - 1.5 * iqr
max_bound = q75 + 1.5 * iqr

# Filter the DataFrame to remove rows where the 'depth' value falls outside the defined bounds.
df_clean = df[(df[col] >= min_bound) & (df[col] <= max_bound)].copy()

# Apply Exponentially Weighted Moving Average (EWMA) to smooth the 'depth' values.
# alpha=0.35 , deteering the weight of recent observations and aw higher alpha gives more weight to recent data, resulting in less smoothing.
df_clean['depth'] = df_clean['depth'].ewm(alpha=0.35).mean()
```

```python
from IPython.display import HTML # Used to display HTML content, specifically for rendering the animation in the notebook.

# --- Visualization Setup ---

# Create a figure and an axes object for the plot & Figure size is set for better visibility.
fig, ax = plt.subplots(figsize=(10, 6))

# Initialize the line object that will be updated in the animation.
# It starts with no data but defines the color, linewidth, and label.
line_filter, = ax.plot([], [], color='blue', linewidth=2, label='Filtered Depth')

# Add a horizontal dashed red line to represent the grounding limit at 20 meters.
# This serves as a critical safety threshold.
ax.axhline(20.0, color='red', linestyle='--', label='Grounding Limit (20m)')

# Setting the labels for the x and y axes to improve readability.
ax.set_xlabel('Point')
ax.set_ylabel('Depth (m)')

# Setting the title of the plot.
ax.set_title('Live Ship Depth Monitor')

# Setting the x-axis limits. based upon the max value
# ensuring all data points are visible, or at least 1 if no data.
ax.set_xlim(0, max(df_clean['point'].max(), 1))

# Setting the y-axis limits to invert it: 0 at the top, max depth at the bottom.
max_y_val = max(df_clean['depth'].abs().max() * 1.05, 25)
ax.set_ylim(max_y_val, 0)

# Display the legend
ax.legend(loc='upper right')


# --- Animation Function ---

# The update function is called for each frame of the animation.
# It takes the current frame number as input.
def update(frame):
    # Select a subset of the data up to the current frame to simulate 'live' data arrival.
    current = df_clean.iloc[:frame]
    # Update the data for the 'Filtered Depth' line.
    line_filter.set_data(current['point'], current['depth'].abs())
    # Return the updated line object. This is required for `blit=True` optimization.
    return line_filter,


# --- Create and Display Animation ---

# Create the animation object using FuncAnimation.
ani = animation.FuncAnimation(
    fig, # The object to be animate.
    update, # The function that updates the plot for each frame.
    frames=len(df_clean) + 1, # The total number of frames, covering all data points.
    interval=100, # Delay between frames in milliseconds.
    blit=True # Blitting can optimize animation performance by redrawing only the parts that have changed.
)

# Close the figure to prevent it from being displayed as a static plot alongside the animation.
plt.close(fig)

# Convert the animation to an HTML representation  which makes the animation viewable without needing to save it to a file first.

HTML(ani.to_jshtml())
```
